```python
# Calculate Q1 and Q3 for 'score'
score_Q1 = df['score'].quantile(0.25)
score_Q3 = df['score'].quantile(0.75)

# Calculate IQR
score_IQR = score_Q3 - score_Q1

# Calculate lower and upper bounds for outliers
score_lower_bound = score_Q1 - 1.5 * score_IQR
score_upper_bound = score_Q3 + 1.5 * score_IQR

print(f"Q1 (score): {score_Q1}")
print(f"Q3 (score): {score_Q3}")
print(f"IQR (score): {score_IQR}")
print(f"Lower bound for outliers (score): {score_lower_bound}")
print(f"Upper bound for outliers (score): {score_upper_bound}")

# Identify outliers
score_outliers_iqr = df[(df['score'] < score_lower_bound) | (df['score'] > score_upper_bound)]

print(f"\nNumber of outliers detected using IQR for 'score': {len(score_outliers_iqr)}")
display(score_outliers_iqr)
```

The KDE plot for 'score' has been generated, and its skewness is approximately -0.995. Since the distribution is negatively skewed (left-skewed), the Interquartile Range (IQR) method would generally be more suitable for outlier detection than the Z-score method. The IQR method is less sensitive to the shape of the distribution, making it a better choice for non-normally distributed data.

Let's apply the IQR method to the 'score' column. I'll calculate the lower and upper bounds and then identify the outliers based on these bounds.

```python
l = [30, 32, 35, 38, 40, 42, 45, 48, 50, 55, 60, 65, 70, 80, 90, 100, 120, 150, 200, 300, 400, 500, 800, 1000, 1200]

```

```python
df = pd.DataFrame(l)
```

```python
df.columns = ['price']
```

```python
df.sample()
```

```python
sns.kdeplot(data=df , x='price')
```

```python
# Calculate Q1 and Q3 for 'price'
price_Q1 = df['price'].quantile(0.25)
price_Q3 = df['price'].quantile(0.75)

# Calculate IQR
price_IQR = price_Q3 - price_Q1

# Calculate lower and upper bounds for outliers
price_lower_bound = price_Q1 - 1.5 * price_IQR
price_upper_bound = price_Q3 + 1.5 * price_IQR

print(f"Q1 (price): {price_Q1}")
print(f"Q3 (price): {price_Q3}")
print(f"IQR (price): {price_IQR}")
print(f"Lower bound for outliers (price): {price_lower_bound}")
print(f"Upper bound for outliers (price): {price_upper_bound}")

# Count outliers before clipping
outliers_before_clip = df[(df['price'] < price_lower_bound) | (df['price'] > price_upper_bound)]
print(f"\nNumber of outliers detected before clipping: {len(outliers_before_clip)}")

# Apply clipping to handle outliers
df['price'] = df['price'].clip(price_lower_bound, price_upper_bound)

print(f"\n'price' column after IQR-based clipping (first 5 rows):")
display(df.head())
print(f"\nMaximum value after clipping: {df['price'].max()}")
print(f"Minimum value after clipping: {df['price'].min()}")
```

The Interquartile Range (IQR) method was chosen for handling outliers in the 'price' column primarily because the distribution of 'price' is likely skewed, as suggested by the KDE plot and the presence of very high values (up to 1200, with an upper bound of 432.5).

*   **Z-score Method**: This method assumes that the data is normally (or approximately normally) distributed. It relies on the mean and standard deviation, which are highly sensitive to extreme values. If the data is skewed, the mean can be pulled towards the tail, and the standard deviation can be inflated, leading to misidentification of outliers (either too many or too few).

*   **IQR Method**: In contrast, the IQR method is more robust to skewed distributions and extreme values. It relies on quartiles (Q1 and Q3), which are less affected by outliers than the mean. This makes it a more appropriate choice when dealing with non-normally distributed data, which is often the case with real-world financial data like 'price'. The IQR method effectively defines a 'normal' range based on the central 50% of the data, making it suitable for identifying outliers in skewed datasets.

```python
l = [1, 2, 3, 4, 4, 5, 5, 5, 6, 6, 7, 7, 8, 8, 9, 9, 10, 10, 1, 2, 10, 10, 10, 0.5, 9.5]

df = pd.DataFrame(l)

df.columns = ['rating']
df.shape
```

```python
x = 5
x = x/100
lb = df['rating'].quantile(x)
ub = df['rating'].quantile(1-x)
df['rating'] = df['rating'].clip(lb, ub)
print(f"Maximum value after Winsorization: {df['rating'].max()}")
print(f"Minimum value after Winsorization: {df['rating'].min()}")
```

```python
df['rating'].min()
```

Winsorization is often preferred over outright deletion of outliers in situations like this because:

Retention of Data: Deleting outliers can lead to a significant loss of data, especially if there are many outliers or if the dataset is small. Winsorization, however, keeps all data points, only adjusting their extreme values, thus preserving the dataset's size and potentially more information.
Preservation of Relationships: Deleting outliers might distort the underlying relationships between variables, especially if the outliers are meaningful data points rather than errors. Winsorization helps to maintain these relationships by merely reducing the impact of extreme values.
Less Impact on Statistical Power: By retaining data, Winsorization generally has a less negative impact on the statistical power of subsequent analyses compared to deletion methods.
After applying 5% Winsorization to the 'rating' column, the maximum value is now 10.0 and the minimum value is 1.0. This indicates that extreme values beyond the 5th and 95th percentiles have been capped at these respective percentile values, effectively reducing the impact of outliers without removing the data points entirely.

```python
l = [12, 15, 18, 20, 22, 22, 23, 24, 25, 25, 26, 27, 28, 30, 32, 35, 38, 40, 45, 50, 55, 60, 65, 70, 80, 90, 100, 120, 150, 200, 500, 1000, 2000]

df = pd.DataFrame(l)
df.columns = ['rating']
df.shape
```

```python
# --- Z-score Method for Outlier Detection on 'rating' ---
mean_rating = df['rating'].mean()
std_rating = df['rating'].std()

df['zscore_rating'] = (df['rating'] - mean_rating) / std_rating

outliers_zscore = df[abs(df['zscore_rating']) > 3]

print(f"\nZ-score Method (threshold |Z| > 3):")
print(f"Number of outliers detected: {len(outliers_zscore)}")
print("Outliers (Z-score):")
display(outliers_zscore[['rating', 'zscore_rating']])


# --- IQR Method for Outlier Detection on 'rating' ---
q1_rating = df['rating'].quantile(0.25)
q3_rating = df['rating'].quantile(0.75)
iqr_rating = q3_rating - q1_rating

lower_bound_iqr = q1_rating - 1.5 * iqr_rating
upper_bound_iqr = q3_rating + 1.5 * iqr_rating

outliers_iqr = df[(df['rating'] < lower_bound_iqr) | (df['rating'] > upper_bound_iqr)]

print(f"\nIQR Method (threshold 1.5 * IQR):")
print(f"Number of outliers detected: {len(outliers_iqr)}")
print("Outliers (IQR):")
display(outliers_iqr[['rating']])

# --- Comparison and Justification ---

print("\n--- Comparison and Justification ---")
print(f"The Z-score method found {len(outliers_zscore)} outliers.")
print(f"The IQR method found {len(outliers_iqr)} outliers.")

if len(outliers_iqr) > len(outliers_zscore):
    print("\nThe IQR method detected more outliers in this case.")
    print("This is often because the Z-score method assumes a normal distribution, and when the data is skewed (as the 'rating' column appears to be from the KDE plot), the mean and standard deviation can be heavily influenced by extreme values, making it less sensitive to outliers in the tails. The IQR method, being based on quartiles, is more robust to skewed distributions.")
elif len(outliers_zscore) > len(outliers_iqr):
    print("\nThe Z-score method detected more outliers in this case.")
    print("This could happen if the distribution has a few extreme values that pull the mean and standard deviation significantly, but the bulk of the data is still within the IQR bounds. However, given the nature of the 'rating' column, extreme high values are more common.")
else:
    print("\nBoth methods detected the same number of outliers.")

print("\nWhich method would I trust here? Justify.")
print("Given the previous KDE plot of the 'rating' column, which showed a right-skewed distribution with a long tail (many high values), the **Interquartile Range (IQR) method** is generally more trustworthy.")
print("\n**Justification:**")
print("1.  **Robustness to Skewness:** The IQR method uses quartiles (Q1 and Q3) which are less sensitive to extreme values than the mean and standard deviation used in the Z-score method. For skewed data, the Z-score's assumption of normality is violated, leading to potentially misleading results.")
print("2.  **Nature of the Data:** In datasets where extreme values are expected to occur in one direction (e.g., very high ratings/prices), the IQR method provides a more stable and reliable boundary for defining outliers. The Z-score might classify values as 'normal' simply because the mean and standard deviation are inflated by the extreme values.")
print("Therefore, for the 'rating' column with its observed skewed distribution, the IQR method provides a more appropriate and reliable identification of outliers.")
```

```python
# 1. IQR Method Outlier Detection
q1 = df['bonus'].quantile(0.25)
q3 = df['bonus'].quantile(0.75)
iqr = q3 - q1
lb_iqr = q1 - 1.5 * iqr
ub_iqr = q3 + 1.5 * iqr

outliers = df[(df['bonus'] < lb_iqr) | (df['bonus'] > ub_iqr)]
print(f"Number of IQR outliers detected: {len(outliers)}")
print("Outlier values:", outliers['bonus'].values)

# 2. Winsorization (x = 3%)
x = 3 / 100
lb_win = df['bonus'].quantile(x)
ub_win = df['bonus'].quantile(1 - x)

df['bonus_winsorized'] = df['bonus'].clip(lb_win, ub_win)

# 3. Show new max and min
print(f"\nNew Max after 3% Winsorization: {df['bonus_winsorized'].max()}")
print(f"New Min after 3% Winsorization: {df['bonus_winsorized'].min()}")
```
