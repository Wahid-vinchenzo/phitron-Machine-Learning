# 0) Starter Code (Run First)

```python
# You must run this code and use this dataset for all questions
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)
num_students = 50

# Student data
student_ids = np.arange(201, 201 + num_students)
student_ages = np.random.randint(15, 22, num_students)
student_scores = np.random.randint(40, 100, num_students)
student_attendance = np.random.randint(60, 100, num_students)  # percentage

departments_list = [
    "Science",
    "Commerce",
    "Arts",
    "Computer Science",
    "Mathematics",
    "English",
    "Physics",
    "Chemistry",
    "Biology",
    "Economics"
]

student_departments = np.random.choice(departments_list, num_students)

grade_conditions = [
    student_scores >= 80,
    (student_scores >= 65) & (student_scores < 80),
    (student_scores >= 50) & (student_scores < 65),
    student_scores < 50
]
grade_values = ["A", "B", "C", "F"]
student_grades = np.select(grade_conditions, grade_values, default="F")

# Create DataFrame
data = pd.DataFrame({
    "student_id": student_ids,
    "student_age": student_ages,
    "score": student_scores,
    "attendance": student_attendance,
    "department": student_departments,
    "grade": student_grades
})

# Save dataset
data.to_csv("school_data.csv", index=False)
```

##
Using .loc:

* Increase the score by 10% for students in "Computer Science" and "Mathematics" departments
* Increase the score by 5% for all other departments
* Cap the score at 100 (no score should exceed 100 after the raise)

Then print a DataFrame that includes:

* student_id, department, previous_score, new_score, score_increase


## Write Answer 04:

```python
#using .loc Increase the score by 10% for students in "Computer Science" and "Mathematics" departments and store tit in new_score
df.loc[(df['department']=='Computer Science') | (df['department'] =='Mathematics'),'new_score'] = df['score'] * 1.1

df.head(3)
```

```python
# using .loc Increase the score by 5% for all other departments
df.loc[df['new_score'].isna(), 'new_score'] = df['score'] * 1.05

df.isnull().sum()
```

```python
df['new_score'].max()
```

```python
# Cap the score at 100
df['new_score'] = df['new_score'].clip(upper=100)
```

```python
df['new_score'].max()
```

```python
# Calculating score_increase
df['score_increase'] = df['new_score'] - df['score']

df.columns
```

```python
# Print the DataFrame with given condition

print(df[['student_id', 'department', 'score', 'new_score', 'score_increase']])
```

# Question 05: [ Marks 10 ]

Using Matplotlib, create a bar chart that shows the average score per department.

* X-axis: Department names
* Y-axis: Average score
* Add a title: "Average Score by Department"
* Display the chart

## Answer 05:

```python
#creating bar chart  that shows average score per department

condition = df.groupby('department')['score'].mean()
condition
```

```python
plt.figure(figsize=(15, 6))
plt.bar(condition.index, condition.values,color='skyblue')
plt.title('Average Score by Department',color ='blue')
plt.xlabel('Department')
plt.ylabel('Average Score')
plt.show()
```
